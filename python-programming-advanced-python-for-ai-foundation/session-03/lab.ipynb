# Library Management System with OOP

## Story: A Real Library

Now, the library has hundreds of books. You need a system to:
- Store all books
- Search for a book quickly
- Track which books are borrowed
- Handle borrowing and returning

This is exactly what **Object-Oriented Programming** helps us do. We create a `Book` class to represent each book, and a `Library` class to manage the entire collection.
