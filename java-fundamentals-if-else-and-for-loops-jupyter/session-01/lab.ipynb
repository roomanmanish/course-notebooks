### Optional self-check (provided for you)

The helper below tries to compile and run whatever is in `JAVA_CODE`. If this environment has no Java compiler it simply says so - that is fine, it is not graded.

````python
import pathlib
import shutil
import subprocess


def try_run_java(stdin_text=""):
    """Compile and run JAVA_CODE if a JDK is available. Not graded."""
    if not JAVA_CODE.strip():
        print("JAVA_CODE is still empty.")
        return
    if shutil.which("javac") is None or shutil.which("java") is None:
        print("No Java compiler in this environment - skip this check.")
        return
    pathlib.Path("Main.java").write_text(JAVA_CODE)
    build = subprocess.run(["javac", "Main.java"], capture_output=True, text=True, timeout=120)
    if build.returncode != 0:
        print(build.stderr)
        return
    run = subprocess.run(["java", "Main"], input=stdin_text, capture_output=True, text=True, timeout=120)
    print(run.stdout or run.stderr)


try:
    try_run_java("7\n85\n")
except Exception as error:
    print("Self-check could not finish:", error)
````
