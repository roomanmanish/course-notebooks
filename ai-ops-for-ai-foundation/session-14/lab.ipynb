# Deploy Agent as a REST API

## Problem

You are building a simple AI agent that answers questions about a knowledge base. Your task is to package this agent as a **REST API** using FastAPI, so it can be called over HTTP by other applications.

The agent should:
- Accept user questions via an HTTP POST endpoint
- Return structured responses (the agent's answer + metadata)
- Handle errors gracefully (invalid input, missing fields, server errors)
- Log all requests and responses for monitoring
- Limit how many requests a user can make in a time window (rate limiting)

Your API will run locally, and you'll test it by sending HTTP requests and checking the responses.

### Real-world context

Imagine you've built a chatbot that helps customers understand a company's product. Right now it only works in a Python script. To let the website, mobile app, and other services use it, you need to expose it as an API—a standardized way for any client to send a question and get an answer back. You also need to log what's happening (for debugging) and prevent one user from hammering your server with thousands of requests per second (rate limiting).

## Expected Output

### Example 1: Successful question
**Request:**
```
POST /ask
Content-Type: application/json

{"question": "What is the capital of France?"}
```

**Response (HTTP 200):**
```json
{
  "question": "What is the capital of France?",
  "answer": "Paris",
  "status": "success"
}
```

### Example 2: Missing question field
**Request:**
```
POST /ask
Content-Type: application/json

{}
```

**Response (HTTP 400):**
```json
{
  "error": "Missing 'question' field",
  "status": "error"
}
```

### Example 3: Rate limit exceeded
**Request:** (5th request within 1 minute from the same client)
```
POST /ask
Content-Type: application/json

{"question": "Another question?"}
```

**Response (HTTP 429):**
```json
{
  "error": "Rate limit exceeded. Max 4 requests per minute.",
  "status": "error"
}
```

### Example 4: Health check endpoint
**Request:**
```
GET /health
```

**Response (HTTP 200):**
```json
{
  "status": "healthy",
  "service": "agent-api"
}
```
