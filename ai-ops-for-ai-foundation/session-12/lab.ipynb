# Custom Tools for a Math Agent

## Overview

In this lab, you will build a LangChain agent equipped with custom tools. Think of tools as specialized helpers that an AI agent can call when it needs to do something it cannot do alone.

**Story**: Imagine you are a student working on a physics homework problem. The question is: *"If I have 3 kilograms of water and I heat it by 20 degrees Celsius, how much energy in kilowatt-hours do I need?"*

You cannot solve this from memory. You need to:
1. **Look up** the specific heat capacity of water (a physical constant)
2. **Calculate** the energy using the formula: Energy = mass × specific heat × temperature change
3. **Convert** the result from joules to kilowatt-hours

An AI agent works the same way. It cannot do all of this alone, so it uses **tools**—functions it can call to get data, perform calculations, or convert units.

## Your Task

Then, create an agent that uses these tools to answer complex, multi-step questions.
