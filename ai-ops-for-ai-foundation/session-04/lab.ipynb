# Math Utility Functions Library

## Story: Building Your First Reusable Code

Imagine you are a baker. Every time you bake a cake, you follow the same steps: mix flour, add eggs, bake at 350°F. Instead of writing down these steps every single time, you write a **recipe card** once. Now, whenever you need to bake, you just follow that one card.

In programming, **functions** are like recipe cards. You write the instructions once, and then you can use them over and over. This saves time, reduces mistakes, and makes your code easier to understand.

In this lab, you will write four math functions that perform common operations. Each function is a small, reusable tool in your programming toolkit.
