# Capstone: Production AI Agent Project

## Your Mission

Design and build a **complete, production-ready AI agent** that solves a real-world problem.

Think of an AI agent like a helpful colleague who can:
- **Understand** what you ask (natural language input)
- **Decide** what to do (reason about the problem)
- **Act** by calling tools (functions that extend its capabilities)
- **Report back** with a clear answer

### Real-World Story

Imagine you run a small online store. Every day you get customer emails asking the same questions: "When will my order arrive?", "Can I return this?", "Do you have size M in blue?"

Instead of hiring someone to answer each email, you build a **Customer Support Agent** that:
1. Reads the customer's question
2. Decides which tool to use (check order status, look up return policy, search inventory)
3. Calls that tool
4. Gives the customer a helpful answer

This is exactly what you'll build in this capstone—but for *your* chosen domain.

## Expected Output

When someone uses your agent, they should get back a **structured response** like this:

```python
{
    "response": "A helpful answer to the user's question",
    "tools_used": ["tool_name_1", "tool_name_2"],
    "success": True
}
```

### Example 1: Customer Support Agent
```
User: "I want to report a bug in the login page"

    "success": True
}

### Example 2: Data Analysis Agent
```
User: "What is the average age in my dataset?"
