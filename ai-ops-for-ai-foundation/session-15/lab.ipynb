# Multi-Agent Research Assistant

## Overview

In this lab, you will build a **multi-agent system** where specialized agents collaborate to answer research questions.

### The Story

Imagine you're running a research organization. When a complex question arrives, you don't ask one person to do everything. Instead:

1. You send the question to your **Research Team** — they gather information and summarize findings
2. You pass their findings to your **Analysis Team** — they identify patterns, evaluate credibility, and draw insights
3. You send the analysis to your **Writing Team** — they craft a clear, well-structured report

Each team communicates through a **shared message board** (message queue), and a **Project Manager** (coordinator) ensures the work flows smoothly from one team to the next.

This is exactly what a **multi-agent system** does in AI: multiple specialized agents work together, communicate, and hand off work to solve complex problems that no single agent could handle as well alone.

---

## Your Task

Implement a multi-agent research system with the following components:

### 1. **Agent Classes**
Each agent is a specialist with a `process()` method:
- **ResearcherAgent**: Takes a question, returns a summary of research findings
- **AnalystAgent**: Takes research findings, returns analytical insights
- **WriterAgent**: Takes insights, returns a polished final report

### 2. **Message Queue**
A shared communication channel where agents can:
- Add messages (agent sends a message)
- Retrieve messages (agent reads what others sent)
- View message history (for auditing and debugging)

### 3. **Coordinator**
Orchestrates the workflow:
- Sends the research question to the Researcher
- Passes Researcher output to the Analyst
- Passes Analyst output to the Writer
- Collects the final report and message history

### 4. **State Tracking**
Each agent tracks its state (e.g., "idle", "processing", "complete") so the coordinator knows when work is done.

---

## Expected Output

When you run the system with a research question, you should get:
1. **Researcher output**: A summary of findings
2. **Analyst output**: Insights and patterns identified
3. **Writer output**: A final, polished report
4. **Message history**: A list of all messages passed between agents
5. **Agent states**: Confirmation that each agent completed their work

Example:
```
Question: "What are the main causes of climate change?"

Researcher: "Greenhouse gases, deforestation, industrial emissions..."
Analyst: "Primary driver is human activity; natural factors are secondary..."
Writer: "## Climate Change: Causes and Implications\n\nClimate change is primarily..."
Message History: [message1, message2, message3, ...]
Agent States: {'researcher': 'complete', 'analyst': 'complete', 'writer': 'complete'}
```
