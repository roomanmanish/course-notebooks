# Product Inventory System

## Story: The Corner Store Ledger

Imagine you own a small corner store. Every morning, you open your ledger and check:
- **What products do I have?** (product names)
- **How much did each cost me?** (price per unit)
- **How many of each do I have?** (quantity in stock)
- **What is my total stock worth?** (price × quantity, summed for all products)

A customer buys 5 apples. You update the ledger: apples go from 50 to 45.

A product expires and you stop selling it. You cross it out of the ledger.

This is exactly what a **dictionary-based inventory system** does. Instead of a paper ledger, we use Python dictionaries to organize and manage product data.

---

## Technical Concept: Dictionaries as Data Stores

A **dictionary** is a collection of **key-value pairs**. In our inventory:
- **Key** = product name (e.g., "Apple")
- **Value** = product details (price and quantity)

We can store the value as another dictionary (nested), so each product has its own "mini-ledger" with price and quantity.

Example structure:
```python
inventory = {
    "Apple": {"price": 1.50, "quantity": 50},
    "Banana": {"price": 0.75, "quantity": 30}
}
```

Now we can:
- **Access** a product's price: `inventory["Apple"]["price"]` → 1.50
- **Update** quantity: `inventory["Apple"]["quantity"] = 40`
- **Check** if a product exists: `"Apple" in inventory`
- **Remove** a product: `del inventory["Banana"]`
- **Iterate** over all products and calculate totals

This mirrors the shopkeeper's workflow: organized, searchable, and easy to update.
