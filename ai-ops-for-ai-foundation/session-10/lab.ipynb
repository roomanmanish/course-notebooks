# Agent Design Case Study: E-Commerce Customer Support

## Scenario

Imagine you work at an online retailer. Every day, hundreds of customers contact support with different problems:
- A laptop arrived broken
- A package hasn't arrived yet
- The wrong item was shipped
- A customer wants to return something

**The old way (static chatbot):** A simple keyword-matching system:
- If message contains "broken" → reply "We're sorry. Please contact support."
- If message contains "where is my order" → reply "Check your email for tracking."
- If message contains "wrong item" → reply "Please contact support."

Notice the problem? The chatbot gives the same response regardless of context. It doesn't understand *why* the customer is upset or *what they actually need*.

**The new way (AI agent):** An agent that:
1. **Perceives** the customer's situation (what happened, how long ago, how urgent)
2. **Reasons** about the best course of action (is the customer eligible for a refund? Is the order still in transit?)
3. **Acts** by recommending a specific solution (process refund, send tracking, ship replacement)

This lab asks you to build a simple agent that demonstrates this loop.

---

## The Agent Loop: Sense → Reason → Act

### Story: A Customer's Journey

Meet Sarah. She ordered a laptop 45 days ago. It arrived with a cracked screen. She emails support:

> "I ordered a laptop 45 days ago. It arrived broken. I want my money back."

**What a keyword chatbot does:**
- Sees "broken" → sends generic "contact support" message
- Sees "money back" → sends generic "refund policy" link
- Sarah is frustrated because the chatbot didn't actually help

**What an agent does:**

1. **Perception (Sense):** The agent reads Sarah's message and extracts facts:
   - Issue type: damaged product
   - Days since purchase: 45
   - Customer request: refund
   - Urgency: high (customer is upset)

2. **Reasoning (Think):** The agent applies business logic:
   - "Is 45 days within our return window (30–60 days)? Yes."
   - "Is the product damaged? Yes."
   - "Is the customer eligible for a refund? Yes."
   - "What's the best action? Process the refund immediately."

3. **Action (Do):** The agent decides:
   - Recommended action: `process_refund`
   - Details: "Initiate full refund. Send return label."

Sarah gets a helpful, specific response in seconds. No waiting for a human.

---

## Key Differences: Agent vs. Chatbot

| Aspect | Chatbot | Agent |
|--------|---------|-------|
| **Input** | Keywords | Full context (message + business rules) |
| **Processing** | Pattern matching | Perception → Reasoning → Action |
| **Output** | Generic template | Specific decision + reasoning |
| **Adaptability** | Fixed responses | Adapts to context |
| **Example** | "broken" → "contact support" | "broken laptop, 45 days" → "process refund" |
