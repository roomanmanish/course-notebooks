# NumPy Inventory System

## Problem

A retail warehouse needs to manage inventory across multiple storage bins. Currently, they use Python lists and manual calculations, which is slow and error-prone.

Your task is to build a NumPy-based system that can:
1. Create and organize inventory data as NumPy arrays
2. Perform calculations (totals, averages, value)
3. Reshape and filter data to identify low-stock products
4. Generate reports with key statistics

### Real-World Context

Imagine you're the warehouse manager. Every morning, your team scans stock levels in 4 storage bins for 3 different products. Instead of writing down numbers on paper and using a calculator, you want a system that instantly tells you:
- How much total stock we have for each product
- Which products are running dangerously low
- The total value of inventory

NumPy arrays make this fast and reliable. A single operation on an array of 1 million items runs in milliseconds, while a Python loop would take seconds.

### Why NumPy?

- **Speed:** NumPy operations are implemented in C and run 10–100× faster than Python loops
- **Memory:** Arrays use less memory than lists
- **Readability:** `array.sum()` is clearer than `sum(my_list)`
- **Broadcasting:** Perform operations on entire arrays without explicit loops
